```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


regio_lijst = (
    'East Asia & Pacific', 
    'Europe & Central Asia', 
    'Latin America & Caribbean', 
    'Middle East & North Africa', 
    'North America', 
    'South Asia', 
    'Sub-Saharan Africa'
)


query_regio = f"""
SELECT country_name, year, value 
FROM co2_emissions_kt_by_country 
WHERE country_name IN {regio_lijst}
AND year >= 1990
ORDER BY year ASC
"""

df_regio = pd.read_sql(query_regio, engine)

# 3. Visualisatie
plt.figure(figsize=(12, 7))
sns.set_style("whitegrid")


sns.lineplot(data=df_regio, x='year', y='value', hue='country_name', linewidth=3)

plt.legend(title='Wereldregio\'s', bbox_to_anchor=(1.02, 1), loc='upper left', frameon=False)

plt.title('Totale CO2-uitstoot per Wereldregio (1990-heden)', fontsize=15, fontweight='bold')
plt.ylabel('CO2-uitstoot in Kiloton (kt)', fontsize=12)
plt.xlabel('Jaar', fontsize=12)

plt.tight_layout()
plt.show()
```
